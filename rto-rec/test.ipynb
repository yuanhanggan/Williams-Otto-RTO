```python
import numpy as np
tf = 200
dt = 8 
dT = 10
```

```python
# Suppose we define some arbitrary function over tf - 0
# y = sin(x) and it is super smooth 
def fa(t):
    return np.sin(t)
```

```python
# we fix this
t_i = np.arange(0, tf + dT, dT)
# given the function fa(t), we can get y
# we also know the place of the where we want the interpolated values 
```

```python
y_inter = np.interp(t_i, t_i, fa(t_i))
```

```python
import matplotlib.pyplot as plt 
plt.plot(t_i, fa(t_i))
```

```python
plt.plot(t_i, y_inter, 'x')
```
